# MF Market Explorer

Interactive analysis of the emerging_re composite rankings for **10-20 unit multifamily investing**.

**What this notebook lets you do:**
1. Browse top markets by 3 target-optimized composites (HPI-max, ZORI-max, JOINT)
2. Filter for actual population + rent growth
3. See existing MF inventory + new-supply pipeline + age-of-stock per county
4. Validate the model with 2022 predictions vs actual outcomes
5. Build a custom shortlist with your own filter criteria

**Data source:** `data/warehouse/emerging_re.duckdb` (built by dbt)

**Run in order first time; then jump around freely.**

## 0. Setup — connect to warehouse

```python
import duckdb
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

# Anchor to project root so this notebook works from any CWD.
PROJECT_ROOT = Path().resolve()
if PROJECT_ROOT.name == 'notebooks':
    PROJECT_ROOT = PROJECT_ROOT.parent
WAREHOUSE = PROJECT_ROOT / 'data/warehouse/emerging_re.duckdb'
assert WAREHOUSE.exists(), f'Warehouse not found at {WAREHOUSE}. Run `cd dbt && uv run dbt build` first.'

con = duckdb.connect(str(WAREHOUSE), read_only=True)
pd.set_option('display.max_rows', 80)
pd.set_option('display.max_columns', 40)
pd.set_option('display.width', 200)
print(f'Connected: {WAREHOUSE}')

def q(sql: str) -> pd.DataFrame:
    """Shorthand — run SQL against warehouse, return DataFrame."""
    return con.execute(sql).fetchdf()
```

## 1. Universe overview

```python
q('''
  select
    'Full county universe (pop>=50K, HPI+ZORI covered)' as scope,
    count(*) as n
  from main_marts.mart_backtest_current
  union all
  select 'MF-filtered (pop>=250K + MF permits>=500 + renter>=25% + vacancy<=8%)', count(*)
  from main_marts.mart_mf_scored
  union all
  select 'Scoreable 2022 cohort (has realized 24mo returns)', count(*)
  from main_marts.mart_backtest_scores
  where period_end_date = date '2022-12-31'
    and hpi_composite_score is not null
''')
```

## 2. Three composites side-by-side

For each MF-filtered county, we compute three target-optimized composites:
- **HPI-max**: features that empirically predict PRICE appreciation (best for value-add/flippers)
- **ZORI-max**: features that predict RENT growth (best for cash-flow investors)
- **JOINT**: strict intersection — features that predict BOTH (safest bet)

```python
# ⚙️ TUNE ME: pick composite + number of markets to show
COMPOSITE = 'zori'  # 'hpi', 'zori', or 'joint'
TOP_N = 20

rank_col = f'mf_rank_{COMPOSITE}'
score_col = f'{COMPOSITE}_composite_score'
q(f'''
  select
    {rank_col} as rank,
    county_name || ' (' || state_name || ')' as market,
    pop_tier,
    pop_cagr_5yr_pct as pop_5y_pct,
    pop_trend,
    round(mf_cap_from_fmr_pct, 2) as cap_pct,
    {score_col} as composite_score,
    rent_annualized_3yr_pct as rent_3y,
    price_annualized_3yr_pct as price_3y,
    est_bldgs_10_to_19 as bldgs,
    sector_diversity_category as sector_div,
    insurance_risk_category as ins_risk,
    worst_peril_score as worst_peril
  from main_marts.mart_mf_scored
  where {rank_col} <= {TOP_N}
  order by {rank_col}
''')
```

## 3. Growth markets — filter for actual pop + rent growth

```python
# ⚙️ TUNE ME: growth thresholds
MIN_MIGRATION_PER_1K = 2       # net domestic in-migration
MIN_RENT_3Y_CAGR_PCT = 4       # 3yr rent growth
MIN_RENT_5Y_CAGR_PCT = 4       # sustained 5yr (filters out COVID one-offs)
MIN_ZORI_COMPOSITE = 0         # forward prediction must be positive
MIN_POP = 100000               # county population floor (drop 100K → smaller markets)

q(f'''
  select
    feat.county_name || ' (' || feat.state_name || ')' as market,
    feat.pop_latest as pop,
    round(f.zori_annualized_3yr * 100, 1) as rent_3y_pct,
    round(f.zori_annualized_5yr * 100, 1) as rent_5y_pct,
    round(f.hpi_annualized_3yr * 100, 1) as price_3y_pct,
    round(mc.domesticmig_per_1000, 1) as mig_per_1k,
    round(feat.zori_composite_score::decimal, 3) as zori_comp,
    round(feat.hpi_composite_score::decimal, 3) as hpi_comp
  from main_marts.mart_backtest_current feat
  join main_intermediate.int_trailing_appreciation f using (fips)
  join main_marts.mart_county_features mc using (fips)
  where feat.pop_latest >= {MIN_POP}
    and mc.domesticmig_per_1000 > {MIN_MIGRATION_PER_1K}
    and f.zori_annualized_3yr >= {MIN_RENT_3Y_CAGR_PCT / 100.0}
    and f.zori_annualized_5yr >= {MIN_RENT_5Y_CAGR_PCT / 100.0}
    and feat.zori_composite_score > {MIN_ZORI_COMPOSITE}
    and (feat.any_disqualifier is null or feat.any_disqualifier = false)
    and mc.period_end_date = (select max(period_end_date) from main_marts.mart_county_features)
  order by (feat.zori_composite_score + mc.domesticmig_per_1000 / 10.0) desc
  limit 30
''')
```

## 4. MF inventory + supply pipeline per county

For MF-filtered universe: existing 10-19 unit building counts (est.), age of stock (pre-1980 share = value-add vintage), new-permit pipeline pressure, and expected annual trades.

```python
# ⚙️ TUNE ME: sort key
SORT_BY = 'est_bldgs_10_to_19'   # try: 'est_annual_trades_10_to_19', 'pre_1980_share_pct', 'pipeline_pct_of_stock'
SORT_DIR = 'desc'

q(f'''
  select
    mf_rank_zori as zori_rank,
    mf_rank_hpi as hpi_rank,
    county_name || ' (' || state_name || ')' as market,
    est_bldgs_10_to_19 as bldgs_10_19,
    est_annual_trades_10_to_19 as trades_yr,
    pre_1980_share_pct as pre_1980_pct,
    pipeline_pct_of_stock as pipeline_pct,
    round(mf_cap_from_fmr_pct, 2) as cap_pct,
    zori_composite_score as zori_comp,
    hpi_composite_score as hpi_comp
  from main_marts.mart_mf_scored
  where est_bldgs_10_to_19 is not null
  order by {SORT_BY} {SORT_DIR}
  limit 25
''')
```

## 5. Backtest validation — 2022 predicted vs 2024/2025 actual

How well did our composites predict rent/price growth in the 2022→2025 window?

```python
# Quintile analysis — Q1 vs Q5 spread by composite
q('''
  with base as (
    select
      ntile(5) over (order by hpi_composite_rank) as hpi_q,
      ntile(5) over (order by zori_composite_rank) as zori_q,
      ntile(5) over (order by joint_composite_rank) as joint_q,
      hpi_forward_return_36mo * 100 as hpi_ret_3y,
      zori_forward_return_36mo * 100 as zori_ret_3y
    from main_marts.mart_backtest_scores
    where period_end_date = date '2022-12-31'
      and hpi_composite_score is not null
  )
  select
    hpi_q as "Quintile (1=best)",
    round(avg(hpi_ret_3y) filter (where hpi_q = hpi_q)::decimal, 2) as "HPI-max→HPI 3y",
    round(avg(zori_ret_3y) filter (where zori_q = hpi_q)::decimal, 2) as "ZORI-max→ZORI 3y",
    round(avg(hpi_ret_3y) filter (where joint_q = hpi_q)::decimal, 2) as "JOINT→HPI 3y",
    round(avg(zori_ret_3y) filter (where joint_q = hpi_q)::decimal, 2) as "JOINT→ZORI 3y"
  from base
  group by hpi_q
  order by hpi_q
''')
```

```python
# Show top 20 predicted by chosen composite → actual outcomes
# ⚙️ TUNE ME
COMPOSITE = 'joint'  # 'hpi', 'zori', 'joint'
TARGET_HORIZON = '36mo'  # '24mo' or '36mo'
TARGET_METRIC = 'zori'   # 'hpi' or 'zori'

q(f'''
  with base as (
    select
      fips, county_name, state_name,
      {COMPOSITE}_composite_rank as pred_rank,
      {COMPOSITE}_composite_score as pred_score,
      {TARGET_METRIC}_forward_return_{TARGET_HORIZON} as actual_ret,
      rank() over (order by {TARGET_METRIC}_forward_return_{TARGET_HORIZON} desc nulls last) as actual_rank
    from main_marts.mart_backtest_scores
    where period_end_date = date '2022-12-31'
      and {COMPOSITE}_composite_rank is not null
  )
  select
    pred_rank, county_name || ' (' || state_name || ')' as market,
    round(pred_score::decimal, 3) as pred_score,
    round(actual_ret * 100, 1) as actual_ret_pct,
    actual_rank
  from base
  where pred_rank <= 20
  order by pred_rank
''')
```

## 6. Custom shortlist builder

Combine your investment criteria into a filtered, ranked shortlist.

```python
# ⚙️ TUNE ME — YOUR CRITERIA
MIN_CAP_PCT = 7.0             # cap rate floor (yield requirement)
MIN_BLDGS_10_19 = 200         # deal-source depth (200 est. bldgs → ~12 trading/yr)
MAX_PIPELINE_PCT = 10.0       # cap new-supply pressure
MIN_ZORI_COMP = 0             # positive rent-growth prediction
MIN_HPI_COMP = 0              # positive price-growth prediction
MIN_POP_CAGR_5YR_PCT = -0.5   # allow slight decline; set >0 for growing only
ALLOWED_INS_RISK = ['Low', 'Moderate', 'High']  # exclude 'Very High' insurance markets
ALLOWED_SECTOR_DIV = ['Diversified', 'Moderate']  # exclude 'Concentrated'
SORT_BY = 'mf_rank_joint'

allowed_ins = "','".join(ALLOWED_INS_RISK)
allowed_div = "','".join(ALLOWED_SECTOR_DIV)

q(f'''
  select
    {SORT_BY} as rank,
    county_name || ' (' || state_name || ')' as market,
    pop_tier,
    pop_cagr_5yr_pct as pop_5y,
    pop_trend,
    round(mf_cap_from_fmr_pct, 2) as cap_pct,
    round(zori_composite_score::decimal, 3) as zori_comp,
    round(hpi_composite_score::decimal, 3) as hpi_comp,
    round(joint_composite_score::decimal, 3) as joint_comp,
    est_bldgs_10_to_19 as bldgs,
    est_annual_trades_10_to_19 as trades_yr,
    pre_1980_share_pct as vintage_pct,
    pipeline_pct_of_stock as pipeline_pct,
    rent_annualized_3yr_pct as rent_3y,
    price_annualized_3yr_pct as price_3y,
    sector_diversity_category as sector_div,
    insurance_risk_category as ins_risk,
    worst_peril_score as worst_peril
  from main_marts.mart_mf_scored
  where mf_cap_from_fmr_pct >= {MIN_CAP_PCT}
    and est_bldgs_10_to_19 >= {MIN_BLDGS_10_19}
    and pipeline_pct_of_stock <= {MAX_PIPELINE_PCT}
    and zori_composite_score >= {MIN_ZORI_COMP}
    and hpi_composite_score >= {MIN_HPI_COMP}
    and (pop_cagr_5yr_pct is null or pop_cagr_5yr_pct >= {MIN_POP_CAGR_5YR_PCT})
    and insurance_risk_category in ('{allowed_ins}')
    and sector_diversity_category in ('{allowed_div}')
  order by {SORT_BY}
  limit 20
''')
```

## 7. Deep-dive on a specific county

```python
# ⚙️ TUNE ME — county to deep-dive
COUNTY_FIPS = '36067'  # Onondaga County NY (Syracuse) — Micron catalyst
# Other examples:
#   '26163' Wayne MI (Detroit),  '36055' Monroe NY (Rochester)
#   '39113' Montgomery OH (Dayton),  '17097' Lake IL

print('══ COUNTY DEEP-DIVE ══')
detail = q(f'''
  select * from main_marts.mart_mf_scored where fips = '{COUNTY_FIPS}'
''')
if len(detail) == 0:
    print(f'County {COUNTY_FIPS} not in MF-filtered universe.')
else:
    # Display as long-form dict for readability
    row = detail.iloc[0].to_dict()
    for k, v in row.items():
        if pd.notna(v):
            print(f'  {k:40s} {v}')
```

## 8. Score contribution — why is this county ranked this way?

Shows the z-scored contribution of each feature to the composite for a specific county.

```python
COUNTY_FIPS = '36067'  # Onondaga NY

z_data = q(f'''
  select * from main_marts.mart_backtest_current
  where fips = '{COUNTY_FIPS}'
''')

if len(z_data) == 0:
    print(f'County {COUNTY_FIPS} not in backtest universe.')
else:
    row = z_data.iloc[0]
    z_cols = [c for c in z_data.columns if c.startswith('z_')]
    contrib = pd.DataFrame({
        'feature': [c.replace('z_', '') for c in z_cols],
        'z_score': [row[c] for c in z_cols],
    }).dropna().sort_values('z_score', ascending=False)
    
    print(f'County: {row["county_name"]} ({row["state_name"]})\n')
    print(f'HPI composite:   {row["hpi_composite_score"]:+.3f} (rank {row["hpi_composite_rank"]})')
    print(f'ZORI composite:  {row["zori_composite_score"]:+.3f} (rank {row["zori_composite_rank"]})')
    print(f'JOINT composite: {row["joint_composite_score"]:+.3f} (rank {row["joint_composite_rank"]})')
    print(f'\nFeature z-scores (higher = stronger positive contribution):')
    print(contrib.to_string(index=False))
```

## 9. Compare two markets side-by-side

```python
# ⚙️ TUNE ME — two counties to compare
FIPS_A = '36067'  # Onondaga NY (Syracuse)
FIPS_B = '39113'  # Montgomery OH (Dayton)

compare_cols = [
    'county_name', 'state_name', 'pop_latest', 'pop_tier',
    'mf_rank_hpi', 'mf_rank_zori', 'mf_rank_joint',
    'mf_cap_from_fmr_pct', 'hpi_composite_score', 'zori_composite_score',
    'est_bldgs_10_to_19', 'est_annual_trades_10_to_19', 'pre_1980_share_pct',
    'pipeline_pct_of_stock',
    'rent_annualized_3yr_pct', 'rent_annualized_5yr_pct',
    'price_annualized_3yr_pct', 'price_annualized_5yr_pct',
    'var_domesticmig_per_1000', 'renter_share_pct', 'vacancy_pct',
    'median_household_income', 'median_home_value', 'median_gross_rent',
]

cmp = q(f'''
  select {','.join(compare_cols)}
  from main_marts.mart_mf_scored
  where fips in ('{FIPS_A}', '{FIPS_B}')
''').T
cmp.columns = ['A', 'B']
cmp
```
